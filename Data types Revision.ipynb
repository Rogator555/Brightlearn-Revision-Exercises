# Data Types:

Text type is a string (these are objects on a dataframe).

Str 

Shown by inverted commas: "My name is Rogator"

Numeric types are integers

int: Age = 29

float: Price = 95.99

Complex: imaginery numbers, often 1i, or 2j etc.

Sequence Types:

- List, square brackets [] ---> mutable, ordered, allows duplicates

- Tuples, rounded brackets () ---> unmutable

- Range, numbers e.g. (1,5)

Mapping Types:

Dictionary, curly brackets {}

Dictionaries have a key:value, example {name:"Rogs, age:29}



Boolean:

True (yes) or False (no)

The first letter of True/False has to capital T or capital letter F otherwise you get an error for using small letters.

Collections:

Storage of multiple items in a single variable.

```python
product1 = "laptop"
product2 = "desktop"
product3 = "tablet"

# We can collect all products in 1 variable:

products = ["laptop", "desktop", "tablet"]
```

```python
print(products)
```

```python
# List can have multiple data types

mixed_list = ["laptop", 200, False, 11.5]
print(mixed_list)
```

### How to access a list:

```python
mixed_list[1]
# The 1st value of the list is 0 then 1,2,3,etc. The last value of the list is -1.
```

### Changing a list:

```python
mixed_list[0] = "cellphone"
print(mixed_list)
```

### Appending a list or adding more:

```python
mixed_list.append("charger")
print(mixed_list)

# The items are by default are appended at the back! Ran the code  twice unfortunately
```

### Deleting a list:

```python
mixed_list.pop(-1)
print(mixed_list)
```

```python
mixed_list = ['cellphone', 200, False, 11.5, 'charger']

display(mixed_list)
```
