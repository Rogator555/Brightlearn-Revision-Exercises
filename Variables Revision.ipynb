### What is a Variable?

A variable is a placeholder, it holds information. Below is an example of a name variable name holding a value called 'Rogator'.

A variable is assigned a value by using an equal sign: '='

Rules of variables:

- always start with a letter or underscore 

Legal variables: name

                :_name

                :_name_surname

                :name10

- you cannot start a variable name with a number; you cannot have a space on the name of the variable; you cannot separate the words in a variable name with a dash rather use an underscore.

Illegal variables

                : 2names = "Rogator Rogs"

                : name surname = "Rogator Monama"

                :name-surname = "Rogator Monama"


Print() function is used to display what is contained in a variable

```python
name = "Rogator"
```

```python
print(name)
```

```python
x = 5

print(x)
```

```python
surnames = ["Mohau", "Mandela", "Dollar", "Grootboom"]

print(surnames)
```

A variable stores different data types. It can store an integer/float (number); strings/objects (shown in inverted commas always, it's a combination of characters); dataframes etc.
